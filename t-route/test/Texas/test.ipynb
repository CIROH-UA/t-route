This jupyter notebook displays the usage of the BMI troute model. If you want to look into the source code, please refer to https://github.com/CIROH-UA/t-route/blob/EKF_Troute/src/troute-nwm/src/nwm_routing/bmi_troute.py 

```python
import sys
import os

# Add the 'src' directory inside 'troute-nwm' to your Python path
sys.path.append(os.path.abspath("../../src/troute-nwm/src"))
from nwm_routing.bmi_troute import BmiTroute
```

```python
#initializing the model
model = BmiTroute()
model.initialize(config_file="troute.yaml")
```

```python
#print the start date time of the model
model.get_current_datetime()
```

```python
#advance the model 15 minutes into the future
model.update()
```

```python
#get reaches from the model
model.get_value_ptr("reach_list")
```

```python
#get discharge value from the model. the indexing is associated with the reach. 
model.get_value_ptr("discharge")
```

```python
#get k value from the model. the indexing is associated with the reach. 
model.get_value_ptr("k")
```

```python
#get x value from the model. the indexing is associated with the reach. 
model.get_value_ptr("x")
```

```python
#check what the current date time is
model.get_current_datetime()
```

```python
#update the model using update_until for a day. the functional argument should be time in seconds
model.update_until(24*60*60)
```

```python
print(model.get_current_datetime())
```

```python
#update for 45 minutes
model.update_until(45*60)
```

This coding snippet below checks whether the output produced by the source code matches the output produced by the BMI model.

```python
import xarray as xr
import numpy as np

output_ds = xr.open_dataset("outputs/troute_output_202208220000.nc")
feature_id = 2431095
feature_id_idx = np.where(model.get_value_ptr("reach_list")==feature_id)
discharge_value_BMI_model = model.get_value_ptr("discharge")[feature_id_idx]
discharge_value_troute_model = output_ds.sel(feature_id=feature_id, time='2022-08-23T01:00:00.000000000').flow.values
print(f"The output value produced by the troute BMI model at time step {model.get_current_datetime()} for feature_id {feature_id} is {discharge_value_BMI_model[0]}")
print(f"The output value produced by the troute source code at time step {model.get_current_datetime()} for feature_id {feature_id} is {discharge_value_troute_model}")
```
